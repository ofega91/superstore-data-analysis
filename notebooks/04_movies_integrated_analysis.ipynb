```python
top5_countries = movies['country'].value_counts().head(5).index

movies_top5_countries = movies[movies['country'].isin(top5_countries)]
```

```python
action = movies[movies['genre'] == 'Action']['gross_millions'].dropna()
action.head()
```

```python
comedy = movies[movies['genre']== 'Comedy']['gross_millions'].dropna()
print(comedy.head(5))
```

```python
from scipy import stats

alpha = 0.05

result = stats.ttest_ind( action, comedy, equal_var = False)

print(result)
```

```python
print('Action mean:', action.mean())
print('Comedy mean:', comedy.mean())
```

### Conclusions

- The dataset is strongly dominated by movies from the United States.
- Budget shows a strong positive relationship with gross, while score has a much weaker relationship.
- Movies with more votes also tend to have higher gross, although this does not imply causation.
- Average gross varies across genres and countries.
- Action movies have a substantially higher average gross than Comedy movies.
- Welch’s t-test indicates that the difference in average gross between Action and Comedy is statistically significant at α = 0.05.
