```python
def calcula_promedio ( puntuaciones):
    try:
        if len(puntuaciones) == 0:
            return None
    
        avg= sum(puntuaciones)/len(puntuaciones)
    
        return avg
    except TypeError:
        return "Datos no Validos"
```

```python
calcula_promedio ([8,6,7,21,5,1,87,12,85])
```

```python
calcula_promedio(0)
```

```python
def convertir_precio(num):
    try:
            nm= float(num)
            return round(nm,2)


    except (ValueError, TypeError):
            
        return None                                  
```

```python
convertir_precio(25.9)
convertir_precio('ocho')
```

```python
convertir_precio(None)
```

```python
convertir_precio("25.987")
```

```python
precios = ["25.987", "18.50", "sin dato", None, "42"]
precios_limpios = []

for mun in precios:
    m = convertir_precio(mun)
    precios_limpios.append(m)
print(precios_limpios)
```

```python
precios_validos = []
for pre in precios_limpios:
    if pre is not None:
        precios_validos.append(pre)

print(precios_validos)
```

```python
precios_promedio = sum(precios_validos)/ len(precios)
print(round(precios_promedio,2))
```

```python
precio_promdio= []
for t in precios_validos:
        if t is not None: 
            precio_promdio.append(t)
```

```python
def promedio_precios1(precios):
    precios_validos = []

    for precio in precios:
        try:
            numero = float(precio)
            precios_validos.append(numero)
        except (ValueError, TypeError):
            continue

    if len(precios_validos) == 0:
        return None

    promedio = sum(precios_validos) / len(precios_validos)
    return round(promedio, 2)
```

```python
print(promedio_precios1(["25.987", "18.50", "sin dato", None, "42"]))
print(promedio_precios1(["sin dato", None]))
print(promedio_precios1(["0", "10"]))
print(promedio_precios1([]))
```

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def calcular_total(self, cantidad):
        return self.precio * cantidad
```

```python
producto = Producto("Cuaderno", 5000)

print(producto.nombre)
print(producto.calcular_total(3))
```
